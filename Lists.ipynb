Find the largest element.

```python
elements = [1, 2, 3, 4, 5, 6, 100, 1, 1000]

largest = elements[0]

for element in elements:
    if element > largest:
        largest = element

print(largest)
```

Find the smallest element.

```python
elements = [1, 2, 3, 4, 5, 6, 100, 1, 1000]

smallest = elements[0]

for element in elements:
    if element < smallest:
        smallest = element

print(smallest)
```

Find the second largest element.

```python
elements = [1, 2, 3, 4, 5, 6, 100, 1, 1000]

largest = elements[0]
second_largest = elements[0]

for element in elements:
    if element > largest:
        second_largest = largest
        largest = element
    elif element > second_largest and element != largest:
        second_largest = element

print("Largest:", largest)
print("Second Largest:", second_largest)
```

Find the average of a list

```python
elements = [1, 2, 3, 4, 5, 6, 100, 1, 1000]

total = 0

for element in elements:
    total += element

average = total / len(elements)

print(average)
```

 Count even and odd numbers

```python
elements = [1, 1, 1, 2, 2, 2, 3, 3, 3, 4, 4, 4, 5, 5, 5, 6, 6, 6]

even_count = 0
odd_count = 0

for element in elements:
    if element % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print("Even:", even_count)
print("Odd:", odd_count)
```

 Sort a list (without using sort()).

```python
elements = [1, 2, 3, 4, 10, 9, 8, 7, 6, 5]

n = len(elements)

for i in range(n - 1):
    for j in range(n - 1 - i):
        if elements[j] > elements[j + 1]:
            temp = elements[j]
            elements[j] = elements[j + 1]
            elements[j + 1] = temp

print(elements)
```

Merge two lists.

 Find common elements.

```python
elements = [1, 2, 3,2,2,3,4]
common=[]
for i in range(len(elements)):
    for j in range(i + 1, len(elements)):
      if elements[i] == elements[j]:
        if elements[i] not in common:
           common.append(elements[i])

print(common)
```

 Find the frequency of each element.

```python
elements = [1, 2, 3,2,2,3,4,4,4,4]
printed = []

for element in elements:

    if element not in printed:

        count = 0

        for value in elements:
            if value == element:
                count += 1

        print(element, count)
        printed.append(element)
```

 Check if a list is sorted.

```python
elements = [1, 2, 3, 4, 10, 9, 8, 7, 6, 5]
copy_elements=elements.copy()
n = len(elements)

for i in range(n - 1):
    for j in range(n - 1 - i):
        if elements[j] > elements[j + 1]:
            temp = elements[j]
            elements[j] = elements[j + 1]
            elements[j + 1] = temp

if copy_elements == elements:
  print("sorted")
else:
 print("Not Sorted")
```

Move all zeros to the end.

```python
elements = [0, 1, 1, 1, 0, 0, 0]

zero_count = 0
result = []

for element in elements:
    if element == 0:
        zero_count += 1
    else:
        result.append(element)

for _ in range(zero_count):
    result.append(0)

print(result)
```

Find missing numbers in a sequence

```python
numbers = [1, 2, 4, 6, 7, 10]

missing=[]
for i in range(numbers[0], numbers[-1] + 1):
  if i not in numbers:
    missing.append(i)
print(missing)
```
